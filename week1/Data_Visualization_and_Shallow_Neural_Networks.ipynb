# Background

Quantities in the world have relationships that can be expressed as functions.  For example, your BMI is equal to $\frac{h}{w^2}$, and the population of the US next year is equal to this year's population $\times \; (1 + \text{growth rate} - \text{death rate})$.

However, some functions—especially the most interesting functions—are hard to learn.  For example, the mapping from the set of photons that hit your retina to $\{cat,\; not\; cat\}$ is a function \[and it's a function incredibly important to computer vision\], but it's hard to represent mathematically.  Not only do we not know what the form of the equation would even be, but theoretically hand-solving for all the parameters would be impossible.  

Therefore, to learn these functions, we use **neural networks**.

## Iris: Modeling

Let's try using **logistic regression** (a shallow neural network) to fit the data!

As a reminder, logistic regression works as follows:

- We take a linear combination of the input values: $z = w^T x = w_0 x_0 + w_1 x_1$
- We then pass it through the sigmoid function: $y = \sigma(z)$, which returns a value between 0 and 1.

Here, what we're mathematically doing is trying to predict a probability between 0 and 1.  For the purpose of classification, our cutoff point is going to be 0.5.  (So if we predict a value $> 0.5$, we predict *Versicolor*, otherwise *Virginica*.)
